```python
# ========================================================================================
# MARKETPLACE REVIEW INTELLIGENCE
# Notebook 01 — Ingestão e Diagnóstico
# ========================================================================================
# Objetivo: Carregar os dados brutos, unificar os dois arquivos JSON 
# e gerar um diagnóstico completo da qualidade dos dados antes de qualquer transformação.
# ========================================================================================

import pandas as pd
import json
import os
from pathlib import Path

# Define o caminho raiz do projeto de forma dinâmica
# Isso garante que funciona independente de onde o projeto está salvo
ROOT = Path().resolve().parent
RAW_PATH = ROOT / 'data' / 'raw'
PROCESSED_PATH = ROOT / 'data' / 'processed'

# Garante que a pasta processed existe
PROCESSED_PATH.mkdir(parents=True, exist_ok=True)

print("✅ Bibliotecas carregadas com sucesso")
print(f"📁 Pasta root: {ROOT}")
print(f"📁 Pasta raw: {RAW_PATH}")
print(f"📁 Pasta processed: {PROCESSED_PATH}")
```

```python
# ============================================================
# Carrega os dois arquivos JSON e unifica em um único DataFrame
# ============================================================
# Substitua os nomes abaixo pelos nomes reais dos seus arquivos

ARQUIVO_1 = RAW_PATH / "reviews_mercadolivre_com_br_1.json"
ARQUIVO_2 = RAW_PATH / "reviews_mercadolivre_com_br_2.json"

def carregar_json(caminho):
    """
    Lê um arquivo JSON com lista de objetos [{...}, {...}]
    e retorna um DataFrame pandas.
    """
    with open(caminho, encoding="utf-8") as f:
        dados = json.load(f)
    return pd.DataFrame(dados)

# Carrega cada arquivo separadamente
df1 = carregar_json(ARQUIVO_1)
df2 = carregar_json(ARQUIVO_2)

print(f"📄 Arquivo 1: {df1.shape[0]:,} linhas | {df1.shape[1]} colunas")
print(f"📄 Arquivo 2: {df2.shape[0]:,} linhas | {df2.shape[1]} colunas")

# Unifica os dois em um único DataFrame
df = pd.concat([df1, df2], ignore_index=True)

print(f"\n✅ Dataset unificado: {df.shape[0]:,} linhas | {df.shape[1]} colunas")
```

```python
# ============================================================
# Visão geral do dataset unificado
# ============================================================

print("=" * 50)
print("PRIMEIRAS 10 LINHAS")
print("=" * 50)
display(df.head(50))

print("\n" + "=" * 50)
print("TIPOS DE DADOS POR COLUNA")
print("=" * 50)
print(df.dtypes)

print("\n" + "=" * 50)
print("VALORES NULOS POR COLUNA")
print("=" * 50)
nulos = df.isnull().sum()
pct_nulos = (nulos / len(df) * 100).round(2)
diagnostico_nulos = pd.DataFrame({
    "Nulos": nulos,
    "% do Total": pct_nulos
})
print(diagnostico_nulos)

print("\n" + "=" * 50)
print("DUPLICATAS EXATAS")
print("=" * 50)
total_duplicatas = df.duplicated().sum()
print(f"Total de linhas duplicadas: {total_duplicatas:,}")
print(f"Percentual: {(total_duplicatas / len(df) * 100):.2f}%")

```

```python
# ============================================================
# Analisa a distribuição dos ratings (1 a 5)
# ============================================================

print("=" * 50)
print("DISTRIBUIÇÃO DOS RATINGS")
print("=" * 50)

dist_rating = df["rating"].value_counts().sort_index() # Quantas vezes cada valor de rating aparece no DataFrame.
pct_rating = (dist_rating / len(df) * 100).round(2) # Qual a porcentagem de cada valor em relação ao total de registros.

diagnostico_rating = pd.DataFrame({
    "Quantidade": dist_rating,
    "% do Total": pct_rating
})
print(diagnostico_rating)

print("\n" + "=" * 50)
print("ESTATÍSTICAS DOS RATINGS")
print("=" * 50)
print(df["rating"].describe()) # Estatísticas descritivas básicas: count, mean, std, min, 25%, 50%, 75%, max.

# Detecta valores fora do range esperado (1 a 5)
fora_range = df[~df["rating"].isin([1, 2, 3, 4, 5])]
print(f"\n⚠️  Ratings fora do range 1-5: {len(fora_range):,} registros")
```

```python
# ============================================================
# Analisa o campo de data — formato PT-BR do Mercado Livre
# ============================================================

print("=" * 50)
print("AMOSTRA DE DATAS ENCONTRADAS")
print("=" * 50)
print(df["date"].value_counts().head(10))

print("\n" + "=" * 50)
print("DATAS NULAS OU VAZIAS")
print("=" * 50)
datas_nulas = df["date"].isnull().sum()
datas_vazias = (df["date"] == "").sum()
print(f"Nulas: {datas_nulas:,}")
print(f"Vazias: {datas_vazias:,}")

print("\n" + "=" * 50)
print("RANGE TEMPORAL (min / max)")
print("=" * 50)
# Conversão temporária só para diagnóstico
meses_pt = {
    "jan": "01", "fev": "02", "mar": "03", "abr": "04",
    "mai": "05", "jun": "06", "jul": "07", "ago": "08",
    "set": "09", "out": "10", "nov": "11", "dez": "12"
}

def converter_data_diagnostico(data_str):
    try:
        partes = data_str.lower().replace(".", "").split()
        return f"{partes[2]}-{meses_pt[partes[1]]}-{partes[0].zfill(2)}"
    except:
        return None

datas_convertidas = df["date"].apply(converter_data_diagnostico)
datas_validas = pd.to_datetime(datas_convertidas, errors="coerce")

print(f"Data mais antiga: {datas_validas.min()}")
print(f"Data mais recente: {datas_validas.max()}")
print(f"Datas inválidas após conversão: {datas_validas.isna().sum():,}")
```

```python
# ============================================================
# Analisa o campo de texto dos reviews
# ============================================================

print("=" * 50)
print("DIAGNÓSTICO DO CONTEÚDO DOS REVIEWS")
print("=" * 50)

# Contagem de palavras por review
df["qtd_palavras"] = df["content"].fillna("").apply(lambda x: len(str(x).split()))

print(f"Reviews nulos:              {df['content'].isnull().sum():,}")
print(f"Reviews vazios (string):    {(df['content'] == '').sum():,}")
print(f"Reviews com menos de 3 palavras: {(df['qtd_palavras'] < 3).sum():,}")
print(f"\nMédia de palavras por review: {df['qtd_palavras'].mean():.1f}")
print(f"Maior review (palavras):      {df['qtd_palavras'].max()}")
print(f"Menor review (palavras):      {df['qtd_palavras'].min()}")

print("\n" + "=" * 50)
print("EXEMPLOS DE REVIEWS MUITO CURTOS")
print("=" * 50)
display(df[df["qtd_palavras"] < 3][["rating", "content"]].head(10))
```

```python
# ============================================================
# Exporta um resumo do diagnóstico em .txt para o GitHub
# Serve como evidência de rigor antes da limpeza
# ============================================================

relatorio = f"""
MARKETPLACE REVIEW INTELLIGENCE
Relatório de Diagnóstico — Dataset Bruto
{'=' * 50}

VOLUME DE DADOS
- Total de registros (unificado): {df.shape[0]:,}
- Arquivo 1: {df1.shape[0]:,} registros
- Arquivo 2: {df2.shape[0]:,} registros

QUALIDADE GERAL
- Duplicatas exatas: {df.duplicated().sum():,} ({(df.duplicated().sum()/len(df)*100):.2f}%)
- Nulos em 'date':        {df['date'].isnull().sum():,}
- Nulos em 'rating':      {df['rating'].isnull().sum():,}
- Nulos em 'content':     {df['content'].isnull().sum():,}
- Nulos em 'product_url': {df['product_url'].isnull().sum():,}

RATINGS
- Distribuição:
{dist_rating.to_string()}
- Fora do range 1-5: {len(fora_range):,}

DATAS
- Data mais antiga:  {datas_validas.min()}
- Data mais recente: {datas_validas.max()}
- Datas inválidas:   {datas_validas.isna().sum():,}

REVIEWS (CONTENT)
- Reviews com menos de 3 palavras: {(df['qtd_palavras'] < 3).sum():,}
- Reviews vazios: {(df['content'].isnull().sum() + (df['content'] == '').sum()):,}
- Média de palavras: {df['qtd_palavras'].mean():.1f}
{'=' * 50}
"""

caminho_relatorio = PROCESSED_PATH / "01_relatorio_diagnostico.txt"
with open(caminho_relatorio, "w", encoding="utf-8") as f:
    f.write(relatorio)

print(relatorio)
print(f"✅ Relatório salvo em: {caminho_relatorio}")

```
